```python
import numpy as np
print("---Simple Attention Machanism Explanation")

encoder_outputs=np.array([
    [0.1,0.2],
    [0.5,0.3],
    [0.2,0.8],
    [0.9,0.1]
])
print("\nEncoder Outputs (4 words, 2 features each):\n{encoder_outputs}")

decoder_hidden_state=np.array([0.7,0.4])
print("\nDecoder Hidden State(what are we looking for):{decoder_hidden_state}")

alignment_score=np.dot(encoder_outputs,decoder_hidden_state)
print(f"\nAlignment Scores (alignment of each word with decoder state):{alignment_score}")

def softmax(x):
  e_x = np.exp(x - np.max(x))
  return e_x / e_x.sum(axis=0)

attention_weights=softmax(alignment_score)
print(f"\nAttention Weights (how much to pay attention to each word):{attention_wieghts}")

print("\nNotice that the word 'student' got a higher weight , meaning its more relevant to learning.")

context_vector = np.sum(encoder_outputs*attention_weights[:, np.newaxis],axis=0)
print(f"\nContext Vector (weighted sum of encoder outputs , paying attention):\n{context_vector}")

```
